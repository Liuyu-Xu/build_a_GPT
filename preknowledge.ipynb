```python
import math
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
def f(x):
    return 3*x**2-4*x+5
```

```python
f(3.0)
```

```python
xs = np.arange(-5, 5, 0.1)
ys = f(xs)
plt.plot(xs, ys)
```
